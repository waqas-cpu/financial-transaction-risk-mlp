### Target Leakage Analysis
In financial fraud detection, target leakage often occurs when features include information that only becomes available *after* a transaction has been flagged or settled (e.g., 'refund_status', 'dispute_id'). Since this dataset (creditcard.csv) contains PCA-transformed features (V1-V28), we will evaluate them based on their statistical relationship and metadata context.

```python
# Target Leakage Analysis Logic

leakage_audit = []

# We check correlation again for fine-grained review
target_corr = df.corr()[target].abs().sort_values(ascending=False)

for col in df.columns:
    if col == target:
        continue

    correlation_value = target_corr[col]

    # Logic for categorization
    if col.startswith('V'):
        # V1-V28 are PCA components. Typically safe if generated from historical data.
        # However, if correlation was > 0.9, we would flag as potential leakage.
        status = "SAFE AT INFERENCE TIME"
        explanation = "PCA transformed feature; represents latent transaction characteristics known at swipe."
    elif col == 'Time':
        status = "SAFE AT INFERENCE TIME"
        explanation = "Seconds elapsed from the first transaction in the dataset; available at runtime."
    elif col == 'Amount':
        status = "SAFE AT INFERENCE TIME"
        explanation = "Transaction value; the primary datum available at inference."
    else:
        status = "UNKNOWN"
        explanation = "Metadata missing; requires domain verification."

    # Hypothetical Leakage Check (Statistical heuristic)
    if correlation_value > 0.8:
        status = "POTENTIAL LEAKAGE"
        explanation += f" CRITICAL: Extremely high correlation ({correlation_value:.2f}) suggests post-facto calculation."

    leakage_audit.append({
        "Feature": col,
        "Status": status,
        "Correlation with Target": round(correlation_value, 4),
        "Exclusion/Inclusion Reasoning": explanation
    })

leakage_report_df = pd.DataFrame(leakage_audit)
display(leakage_report_df)
```

### Preprocessing Decisions for MLP Neural Network

Based on the EDA, we should implement the following steps before training the MLP:

1.  **Duplicate Removal**: The 1,081 duplicates should be dropped to prevent the model from overfitting on exact transaction signatures.
2.  **Amount Transformation**: The `Amount` feature is extremely skewed (skew > 15). We must apply a **Log Transformation** or **Power Transformation** followed by **StandardScaler** to ensure the MLP weights don't explode.
3.  **Time Handling**: `Time` is a relative offset. It should be scaled using **RobustScaler** or transformed into periodic signals if cyclical patterns are found, though here standard scaling is likely sufficient.
4.  **Imbalance Mitigation**: Since fraud is <0.2%, we must use **Class Weights** in the MLP loss function or perform **Synthetic Minority Over-sampling (SMOTE)** on the training set.
5.  **Feature Selection**: The PCA features (V1-V28) show low multi-collinearity, so all should be kept to provide maximum signal for the latent risk classification.

### Data Splitting Strategy: Chronological Partitioning
In financial fraud detection, transaction order matters. Using a random split can lead to **temporal leakage**, where the model learns patterns from future transactions to predict past ones.

Since our dataset contains a `Time` feature (seconds elapsed), we will sort the data chronologically and use a time-based split (70/15/15). This ensures the model is evaluated on its ability to generalize to future unseen transactions.

```python
# 1. Chronological Sorting
# Ensure data is ordered by time before splitting
df_clean = df_clean.sort_values(by='Time').reset_index(drop=True)

# 2. Define Split Indices
total_rows = len(df_clean)
train_end = int(total_rows * 0.70)
val_end = int(total_rows * 0.85)

# 3. Create Splits
train_df = df_clean.iloc[:train_end]
val_df = df_clean.iloc[train_end:val_end]
test_df = df_clean.iloc[val_end:]

# 4. Verify Shapes
print(f"Train shape: {train_df.shape}")
print(f"Validation shape: {val_df.shape}")
print(f"Test shape: {test_df.shape}")
```

```python
# 5. Verify Target Distribution (Class Imbalance) across splits
def get_distribution(data, name):
    dist = data['Class'].value_counts(normalize=True)
    count = data['Class'].value_counts()
    print(f"--- {name} Distribution ---")
    print(f"Legit (0): {dist[0]:.4%} ({count[0]} rows)")
    print(f"Fraud (1): {dist[1]:.4%} ({count[1]} rows)\n")

get_distribution(train_df, "Train")
get_distribution(val_df, "Validation")
get_distribution(test_df, "Test")
```

### Detailed Imbalance Analysis & Class Weight Calculation

In financial fraud detection, the minority class (fraud) is the priority. We must calculate the imbalance ratio and define an initial strategy for the MLP. Accuracy is inappropriate because a model predicting 0 for every transaction would achieve >99% accuracy while missing 100% of the fraud.

```python
from sklearn.utils.class_weight import compute_class_weight
import numpy as np

# 1. Calculate Imbalance Statistics
counts = train_df['Class'].value_counts()
percentages = train_df['Class'].value_counts(normalize=True) * 100
imbalance_ratio = counts[0] / counts[1]

print("--- Training Set Imbalance Report ---")
print(f"Legitimate Transactions: {counts[0]} ({percentages[0]:.4f}%)")
print(f"Fraudulent Transactions: {counts[1]} ({percentages[1]:.4f}%)")
print(f"Imbalance Ratio: {imbalance_ratio:.2f}:1")

# 2. Calculate Class Weights for MLP Loss Function
# This serves as a safe alternative to SMOTE
classes = np.unique(train_df['Class'])
weights = compute_class_weight(class_weight='balanced', classes=classes, y=train_df['Class'])
class_weight_dict = dict(zip(classes, weights))

print(f"\nCalculated Class Weights: {class_weight_dict}")
```

### Strategy Recommendation: Class Weights vs. SMOTE

For a first-pass MLP in a leakage-safe pipeline:
1.  **Class Weights**: Safest. It modifies the importance of each sample during backpropagation without changing the data distribution. Highly recommended for large datasets like this.
2.  **SMOTE**: Effective for extremely small minority counts. Since we have ~384 fraud cases in the training set, SMOTE is viable but carries a risk of generating synthetic samples that do not respect the strict latent boundaries of financial features.

**Evaluation Metric Note**: We will focus on **PR-AUC (Precision-Recall Area Under Curve)**. Unlike ROC-AUC, which can be overly optimistic for highly imbalanced data, PR-AUC focuses on the model's ability to find the few positive cases without flagging too many false positives.

### Why this split is appropriate:
1. **Temporal Integrity**: By using the `Time` column to segment the data, we simulate a production environment where the model is trained on historical data and tested on subsequent transactions.
2. **Prevention of Leakage**: A random split might place a 'fraudulent burst' (multiple transactions from the same card in a short window) into both train and test sets, artificially inflating performance. Chronological splitting avoids this.
3. **Distribution Consistency**: Although not strictly stratified (as we preserved time order), we will monitor the fraud rates to ensure the minority class is sufficiently represented in the validation and test sets for meaningful evaluation.

### Leakage-Safe Preprocessing Pipeline

We will now construct the final feature set. The pipeline will:
* Remove identifiers/non-predictive columns (`Time`).
* Log-transform `Amount` to handle its extreme skewness (16.98).
* Apply `StandardScaler` to all features to ensure zero mean and unit variance for the MLP.

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, FunctionTransformer

# 1. Define feature groups
# We drop 'Time' as per audit and keep V1-V28
v_features = [f'V{i}' for i in range(1, 29)]

# 2. Define custom log transformer for Amount
log_pipeline = Pipeline(steps=[
    ('log', FunctionTransformer(np.log1p, validate=False)),
    ('scaler', StandardScaler())
])

# 3. Create ColumnTransformer
# PCA features (V1-V28) are already centered but standard scaling ensures consistency
preprocessor = ColumnTransformer(
    transformers=[
        ('v_cols', StandardScaler(), v_features),
        ('amt_log', log_pipeline, ['Amount'])
    ],
    remainder='drop'  # Drops 'Time' and 'Class' (Class is our target)
)

# 4. Fit ONLY on training data to prevent leakage
X_train = preprocessor.fit_transform(train_df)
y_train = train_df['Class'].values

# 5. Transform Validation and Test sets using the fitted preprocessor
X_val = preprocessor.transform(val_df)
y_val = val_df['Class'].values

X_test = preprocessor.transform(test_df)
y_test = test_df['Class'].values

print(f"Preprocessing Complete.")
print(f"Final Input Dimensionality: {X_train.shape[1]}")
print(f"Train observations: {X_train.shape[0]}")
```

### Handling Class Imbalance with SMOTE

Given that fraud accounts for only ~0.19% of our training data, a standard MLP might converge to a trivial solution (predicting all samples as legit). We will use SMOTE to oversample the minority class in the training set.

```python
from imblearn.over_sampling import SMOTE
from collections import Counter

print(f"Original training class distribution: {Counter(y_train)}")

# Initialize SMOTE
smote = SMOTE(sampling_strategy='auto', random_state=42)

# Fit SMOTE only on the training data
X_train_res, y_train_res = smote.fit_resample(X_train, y_train)

print(f"Resampled training class distribution: {Counter(y_train_res)}")
print(f"New training shape: {X_train_res.shape}")
```

### MLP Architecture Design

We will now define a deep neural network using TensorFlow/Keras. For financial fraud detection, we typically use:
* **Input Layer**: 29 nodes.
* **Hidden Layers**: Dense layers with ReLU activation and Dropout to prevent overfitting.
* **Output Layer**: 1 node with Sigmoid activation for binary classification.

### MLP Architecture Definition

We build a deep neural network tailored for high-dimensional classification.

* **Input Layer**: 29 features (V1-V28 and Log-Amount).
* **Batch Normalization**: Standardizes activations between layers, reducing internal covariate shift.
* **Dropout**: Randomly deactivates 20-30% of neurons during training to enforce feature redundancy and prevent overfitting.
* **Trainable Parameters**: Calculated based on weights and biases between dense layers.
* **Sigmoid Activation**: Maps the output to a (0, 1) range, representing the probability of fraud.
* **Binary Crossentropy**: The standard loss function for binary classification, measuring the distance between the predicted probability and the ground truth.

```python
import tensorflow as tf
from tensorflow.keras import layers, models, regularizers

def build_mlp(input_dim):
    model = models.Sequential([
        # Input Layer
        layers.Input(shape=(input_dim,)),

        # Hidden Layer 1 (Added L2 Regularization & increased Dropout)
        layers.Dense(128, activation='relu', kernel_regularizer=regularizers.l2(0.001)),
        layers.BatchNormalization(),
        layers.Dropout(0.4),

        # Hidden Layer 2 (Added L2 Regularization & increased Dropout)
        layers.Dense(64, activation='relu', kernel_regularizer=regularizers.l2(0.001)),
        layers.BatchNormalization(),
        layers.Dropout(0.3),

        # Hidden Layer 3
        layers.Dense(32, activation='relu'),
        layers.Dropout(0.2),

        # Output Layer
        layers.Dense(1, activation='sigmoid')
    ])

    model.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
        loss='binary_crossentropy',
        metrics=[
            tf.keras.metrics.AUC(name='pr_auc', curve='PR'),
            tf.keras.metrics.Precision(name='precision'),
            tf.keras.metrics.Recall(name='recall')
        ]
    )
    return model

# Initialize model based on original X_train dimensionality (No SMOTE)
input_dim = X_train.shape[1]
mlp_model = build_mlp(input_dim)

# Print summary and parameter count
mlp_model.summary()
```

### MLP Training with Dynamic Callbacks

We utilize a production-grade training strategy with callbacks to manage the learning rate and prevent overfitting on the resampled data.

```python
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau, ModelCheckpoint
import os

# 1. Define Callbacks
callbacks = [
    EarlyStopping(
        monitor='val_pr_auc',
        mode='max',
        patience=7,
        restore_best_weights=True,
        verbose=1
    ),
    ReduceLROnPlateau(
        monitor='val_loss',
        factor=0.2,
        patience=3,
        min_lr=1e-6,
        verbose=1
    ),
    ModelCheckpoint(
        'best_mlp_model.keras',
        monitor='val_pr_auc',
        mode='max',
        save_best_only=True,
        verbose=0
    )
]

# 2. Execute Training
# Switched back to original X_train, y_train and added class_weight_dict
history = mlp_model.fit(
    X_train,
    y_train,
    validation_data=(X_val, y_val),
    epochs=50,
    batch_size=2048,
    class_weight=class_weight_dict, # <-- Applying the weights calculated earlier!
    callbacks=callbacks,
    verbose=1
)
```

### Convergence Diagnostics

We plot the learning curves for Loss and Precision-Recall AUC to identify the point of convergence and check for variance (overfitting).

```python
import matplotlib.pyplot as plt

def plot_history(history):
    metrics = ['loss', 'pr_auc', 'precision', 'recall']
    fig, axes = plt.subplots(2, 2, figsize=(15, 10))
    axes = axes.flatten()

    for i, metric in enumerate(metrics):
        axes[i].plot(history.history[metric], label=f'Train {metric}')
        axes[i].plot(history.history[f'val_{metric}'], label=f'Val {metric}')
        axes[i].set_title(f'Model {metric.capitalize()}')
        axes[i].set_xlabel('Epoch')
        axes[i].set_ylabel(metric.capitalize())
        axes[i].legend()

    plt.tight_layout()
    plt.show()

plot_history(history)
```

### Final Model Evaluation (Test Set)

We now evaluate the MLP on the test set to obtain a final, unbiased estimate of performance on the most recent transactions.

```python
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score, average_precision_score, precision_recall_curve, roc_curve, ConfusionMatrixDisplay
import matplotlib.pyplot as plt
import numpy as np

# 1. Generate probabilities for the Validation Set
y_val_probs = mlp_model.predict(X_val).flatten()

# 2. Calculate AUC metrics
roc_auc = roc_auc_score(y_val, y_val_probs)
pr_auc = average_precision_score(y_val, y_val_probs)

print(f"Validation ROC-AUC: {roc_auc:.4f}")
print(f"Validation PR-AUC (Average Precision): {pr_auc:.4f}")

# 3. Find optimal threshold based on F1-score from PR curve
precision, recall, thresholds = precision_recall_curve(y_val, y_val_probs)
fscore = (2 * precision * recall) / (precision + recall + 1e-10)
ix = np.argmax(fscore)
best_thresh = thresholds[ix]
print(f"\nBest Threshold based on F1-score: {best_thresh:.4f}")

# 4. Generate predictions using the optimal threshold
y_val_pred = (y_val_probs >= best_thresh).astype(int)

# 5. Calculate metrics at optimal threshold
print("\n--- Classification Report (Optimal Threshold) ---")
print(classification_report(y_val, y_val_pred))
```

### Threshold Analysis on Validation Set
We will evaluate various thresholds (from 0.05 to 0.95) to see how they impact Precision, Recall, F1-Score, False Positives, and False Negatives.

```python
import numpy as np
import pandas as pd
from sklearn.metrics import confusion_matrix, precision_score, recall_score, f1_score

# Define thresholds to evaluate
thresholds = np.arange(0.05, 0.96, 0.05)
results = []

for thresh in thresholds:
    # Generate predictions based on current threshold
    preds = (y_val_probs >= thresh).astype(int)

    # Calculate confusion matrix and metrics
    tn, fp, fn, tp = confusion_matrix(y_val, preds).ravel()
    precision = precision_score(y_val, preds, zero_division=0)
    recall = recall_score(y_val, preds, zero_division=0)
    f1 = f1_score(y_val, preds, zero_division=0)

    results.append({
        'Threshold': thresh,
        'Precision': precision,
        'Recall': recall,
        'F1-Score': f1,
        'False Positives (FP)': fp,
        'False Negatives (FN)': fn
    })

# Create and display DataFrame
df_thresholds = pd.DataFrame(results)
display(df_thresholds.style.background_gradient(cmap='viridis', subset=['F1-Score', 'Recall', 'Precision']).format({'Threshold': '{:.2f}', 'Precision': '{:.4f}', 'Recall': '{:.4f}', 'F1-Score': '{:.4f}'}))
```

### Business Tradeoff: Fraud Recall vs. False Positive Rate

In financial fraud detection, selecting the optimal threshold is a careful balancing act between preventing financial loss and ensuring a smooth customer experience:

*   **Lower Thresholds (e.g., 0.10 - 0.40)**:
    *   **Effect**: The model aggressively flags transactions as fraudulent.
    *   **Pros**: Maximizes **Recall**. Very few False Negatives mean the bank stops almost all fraud, minimizing direct financial losses.
    *   **Cons**: Skyrocketing **False Positives**. Legitimate customers frequently have their cards declined, resulting in frustration, increased customer support costs, and the risk of customers abandoning the card ("top-of-wallet" loss).

*   **Higher Thresholds (e.g., 0.80 - 0.95)**:
    *   **Effect**: The model only flags transactions it is highly confident about.
    *   **Pros**: Maximizes **Precision**. Almost no False Positives, meaning minimal friction for legitimate customers.
    *   **Cons**: Lower **Recall**. False Negatives increase, meaning actual fraud slips through, resulting in chargebacks and direct financial losses.

#### Threshold Recommendation for Financial Risk
For this specific model, while standard accuracy optimizes around 0.50, financial models often require a custom threshold. Based on the evaluation table above, a threshold around **0.80 to 0.90** is highly recommended.

At these higher thresholds, the model retains strong Recall (catching the vast majority of fraud) while drastically reducing the number of False Positives. If the bank values customer experience over tiny marginal fraud losses, you can push the threshold up slightly. If catching every single fraud dollar is paramount regardless of customer friction, you would pull the threshold down towards **0.50** to capture the remaining edge cases at the expense of higher False Positives.

### Post-Hoc Model Explainability with SHAP

While neural networks are often considered "black boxes," we can use **SHAP (SHapley Additive exPlanations)** to interpret their decisions. SHAP assigns each feature an importance value for a particular prediction.

Here we will extract:
1. **Global Explanations**: Which features are most important overall, and do they generally increase or decrease risk?
2. **Local Explanations**: Why did the model make a specific decision for a single transaction? We will look at a True Positive (caught fraud), a False Positive (false alarm), and a False Negative (missed fraud).

*Note: SHAP values show statistical contribution to the model's output, not necessarily real-world causation.*

```python
!pip install -q shap
import shap
import numpy as np
import matplotlib.pyplot as plt

# 1. Reconstruct feature names from our preprocessing pipeline
feature_names = [f'V{i}' for i in range(1, 29)] + ['Amount_Log']

# 2. Define a prediction function wrapper for SHAP
def predict_fn(x):
    return mlp_model.predict(x, verbose=0).flatten()

# 3. Create a background dataset summarizing the training data
# We use k-means to reduce the background data to 50 representative centroids.
# This makes the KernelExplainer computationally feasible while preserving distribution.
print("Initializing SHAP Explainer...")
background = shap.kmeans(X_train, 50)
explainer = shap.KernelExplainer(predict_fn, background)

# 4. Sample the test set for explanations
# We include ALL fraud cases in the test set and a random sample of legitimate ones.
fraud_idx = np.where(y_test == 1)[0]
legit_idx = np.where(y_test == 0)[0]

np.random.seed(42)
sample_legit_idx = np.random.choice(legit_idx, 150, replace=False)
test_sample_idx = np.concatenate([fraud_idx, sample_legit_idx])

X_test_sample = X_test[test_sample_idx]
y_test_sample = y_test[test_sample_idx]
y_test_sample_pred = y_test_pred_mlp[test_sample_idx]  # Using the 0.85 threshold predictions

print(f"Calculating SHAP values for {len(X_test_sample)} test instances (This may take a minute)...")
# Calculate SHAP values
shap_values = explainer.shap_values(X_test_sample, nsamples=100)
```

#### 1. Global Feature Importance
The plots below show which features drive the model's risk scores globally.
* The **Bar Plot** ranks features by absolute impact.
* The **Summary Plot (Beeswarm)** shows directional impact: whether a high/low value of a feature increases or decreases the predicted risk probability.

```python
plt.figure(figsize=(10, 6))
plt.title("SHAP Global Feature Importance (Mean Absolute Impact)")
shap.summary_plot(shap_values, X_test_sample, feature_names=feature_names, plot_type="bar", show=False)
plt.show()

plt.figure(figsize=(10, 6))
plt.title("SHAP Directional Feature Effects")
shap.summary_plot(shap_values, X_test_sample, feature_names=feature_names, show=False)
plt.show()
```

#### 2. Local Explanations (Transaction-Level)

Now, we inspect individual transactions. The **Waterfall Plot** starts from the baseline probability (expected value) and adds/subtracts the SHAP value of each feature to reach the final predicted probability for that specific transaction.

We will examine:
* A **True Positive**: Fraud that the model successfully caught.
* A **False Positive**: A legitimate transaction that caused a false alarm.
* A **False Negative**: A fraudulent transaction that bypassed our filters.

```python
# Identify indices within our SHAP sample for specific confusion matrix quadrants
tp_idx_sample = np.where((y_test_sample == 1) & (y_test_sample_pred == 1))[0]
fp_idx_sample = np.where((y_test_sample == 0) & (y_test_sample_pred == 1))[0]
fn_idx_sample = np.where((y_test_sample == 1) & (y_test_sample_pred == 0))[0]

def plot_waterfall(indices, title, expected_value, shap_vals, data, features):
    if len(indices) == 0:
        print(f"\nNo instances found for: {title}")
        return

    # Pick the first transaction in the category
    idx = indices[0]
    print(f"\n{'='*50}")
    print(f"{title}")
    print(f"Actual Class: {y_test_sample[idx]} | Predicted Risk Probability: {predict_fn(data[idx:idx+1])[0]:.4f}")
    print(f"{'='*50}")

    # Create a SHAP Explanation object for the waterfall plot
    exp = shap.Explanation(
        values=shap_vals[idx],
        base_values=expected_value,
        data=data[idx],
        feature_names=features
    )

    plt.figure(figsize=(8, 5))
    shap.plots.waterfall(exp, max_display=10, show=False)
    plt.title(f"Explanation: {title}")
    plt.tight_layout()
    plt.show()

# The expected value for KernelExplainer is usually a single float for binary classification
base_value = explainer.expected_value
if isinstance(base_value, np.ndarray):
    base_value = base_value[0]

# Generate Local Plots
plot_waterfall(tp_idx_sample, "True Positive (High-Risk Fraud Successfully Caught)", base_value, shap_values, X_test_sample, feature_names)
plot_waterfall(fp_idx_sample, "False Positive (Legitimate Transaction Falsely Flagged)", base_value, shap_values, X_test_sample, feature_names)
plot_waterfall(fn_idx_sample, "False Negative (Fraudulent Transaction Missed)", base_value, shap_values, X_test_sample, feature_names)
```

### Production Inference Pipeline

This section demonstrates how the saved model artifacts can be used in a production environment (like a REST API).

We define a robust `predict_fraud_risk` function that:
1. Takes a raw transaction (Dictionary or DataFrame).
2. Validates that all required columns are present (Schema Validation).
3. Uses the saved `joblib` preprocessor to transform the data exactly as it was during training.
4. Feeds the transformed data into the `.keras` MLP model.
5. Applies the configured threshold from `model_metadata.json` to assign a business label.

```python
import os
import json
import joblib
import pandas as pd
import tensorflow as tf

# 1. Load Artifacts (Simulating what an API startup script would do)
artifact_dir = '/content/artifacts'

print("Loading production artifacts...")
prod_model = tf.keras.models.load_model(os.path.join(artifact_dir, 'mlp_fraud_model.keras'))
prod_preprocessor = joblib.load(os.path.join(artifact_dir, 'preprocessor.joblib'))

with open(os.path.join(artifact_dir, 'model_metadata.json'), 'r') as f:
    prod_metadata = json.load(f)

def predict_fraud_risk(transaction_data):
    """
    Production-style inference function for a single transaction or batch.
    """
    # 1. Standardize input format
    if isinstance(transaction_data, dict):
        df_trans = pd.DataFrame([transaction_data])
    elif isinstance(transaction_data, pd.DataFrame):
        df_trans = transaction_data.copy()
    else:
        raise ValueError("Input must be a Python dictionary or pandas DataFrame.")

    # 2. Schema Validation
    expected_cols = prod_metadata['feature_metadata']['expected_raw_columns']
    missing_cols = [col for col in expected_cols if col not in df_trans.columns]
    if missing_cols:
        raise ValueError(f"Schema Validation Failed. Missing required columns: {missing_cols}")

    # 3. Preprocessing (reuses the fitted pipeline, ignores extra columns like 'Time' automatically)
    try:
        X_processed = prod_preprocessor.transform(df_trans)
    except Exception as e:
        raise RuntimeError(f"Preprocessing pipeline failed: {str(e)}")

    # 4. Model Prediction
    prob = float(prod_model.predict(X_processed, verbose=0)[0][0])

    # 5. Apply Configured Threshold
    threshold = prod_metadata['decision_threshold']
    risk_class = 1 if prob >= threshold else 0
    risk_label = "HIGH_RISK" if risk_class == 1 else "LOW_RISK"

    # 6. Return Structured Output
    return {
        "risk_probability": round(prob, 4),
        "risk_class": risk_class,
        "risk_label": risk_label,
        "threshold": threshold
    }

print("\u2705 Inference pipeline ready.")
```

```python
%%writefile /content/financial-transaction-risk-mlp/src/preprocessing.py
import numpy as np
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, FunctionTransformer

def get_preprocessor():
    """
    Builds and returns the scikit-learn preprocessing pipeline for financial transactions.
    """
    v_features = [f'V{i}' for i in range(1, 29)]

    log_pipeline = Pipeline(steps=[
        ('log', FunctionTransformer(np.log1p, validate=False)),
        ('scaler', StandardScaler())
    ])

    preprocessor = ColumnTransformer(
        transformers=[
            ('v_cols', StandardScaler(), v_features),
            ('amt_log', log_pipeline, ['Amount'])
        ],
        remainder='drop'
    )

    return preprocessor
```

```python
%%writefile /content/financial-transaction-risk-mlp/src/model.py
import tensorflow as tf
from tensorflow.keras import layers, models, regularizers

def build_mlp(input_dim):
    """
    Builds and compiles the MLP Neural Network for fraud detection.
    """
    model = models.Sequential([
        layers.Input(shape=(input_dim,)),

        layers.Dense(128, activation='relu', kernel_regularizer=regularizers.l2(0.001)),
        layers.BatchNormalization(),
        layers.Dropout(0.4),

        layers.Dense(64, activation='relu', kernel_regularizer=regularizers.l2(0.001)),
        layers.BatchNormalization(),
        layers.Dropout(0.3),

        layers.Dense(32, activation='relu'),
        layers.Dropout(0.2),

        layers.Dense(1, activation='sigmoid')
    ])

    model.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
        loss='binary_crossentropy',
        metrics=[
            tf.keras.metrics.AUC(name='pr_auc', curve='PR'),
            tf.keras.metrics.Precision(name='precision'),
            tf.keras.metrics.Recall(name='recall')
        ]
    )
    return model
```

```python
%%writefile /content/financial-transaction-risk-mlp/src/inference.py
import os
import json
import joblib
import pandas as pd
import tensorflow as tf

def load_production_artifacts(artifact_dir='artifacts'):
    """
    Loads the trained model, preprocessor, and metadata from the artifacts directory.
    """
    model = tf.keras.models.load_model(os.path.join(artifact_dir, 'mlp_fraud_model.keras'))
    preprocessor = joblib.load(os.path.join(artifact_dir, 'preprocessor.joblib'))

    with open(os.path.join(artifact_dir, 'model_metadata.json'), 'r') as f:
        metadata = json.load(f)

    return model, preprocessor, metadata

def predict_fraud_risk(transaction_data, model, preprocessor, metadata):
    """
    Takes raw transaction data, preprocesses it, and returns a fraud risk assessment.
    """
    if isinstance(transaction_data, dict):
        df_trans = pd.DataFrame([transaction_data])
    elif isinstance(transaction_data, pd.DataFrame):
        df_trans = transaction_data.copy()
    else:
        raise ValueError("Input must be a Python dictionary or pandas DataFrame.")

    # Preprocess
    X_processed = preprocessor.transform(df_trans)

    # Predict
    prob = float(model.predict(X_processed, verbose=0)[0][0])

    # Classify
    threshold = metadata['decision_threshold']
    risk_class = 1 if prob >= threshold else 0
    risk_label = "HIGH_RISK" if risk_class == 1 else "LOW_RISK"

    return {
        "risk_probability": round(prob, 4),
        "risk_class": risk_class,
        "risk_label": risk_label,
        "threshold": threshold
    }
```
